```python
import pandas as pd

# Load dataset
df = pd.read_csv("Iris.csv")

# Show first 5 rows
df.head()
```

```python
# Features and target
X = df.drop("Species", axis=1)
y = df["Species"]

# Check shapes
print(X.shape)
print(y.shape)
```

```python
from sklearn.model_selection import train_test_split

# Train on 50% of data
X_train, _, y_train, _ = train_test_split(
    X, y, train_size=0.5, random_state=42
)

# Test on full dataset
X_test = X
y_test = y

print(X_train.shape)
print(X_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler

# Initialize scaler
scaler = StandardScaler()

# Fit only on training data
X_train_scaled = scaler.fit_transform(X_train)

# Transform test data
X_test_scaled = scaler.transform(X_test)
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, f1_score

# Create model
knn = KNeighborsClassifier()

# Train model
knn.fit(X_train_scaled, y_train)

# Predict on full dataset
y_pred_knn = knn.predict(X_test_scaled)

# Evaluate
acc_knn = accuracy_score(y_test, y_pred_knn)
f1_knn = f1_score(y_test, y_pred_knn, average='weighted')

print("KNN Accuracy:", acc_knn)
print("KNN F1 Score:", f1_knn)
```

```python
from sklearn.linear_model import LogisticRegression

# Create model
lr = LogisticRegression(max_iter=200)

# Train model (use scaled data)
lr.fit(X_train_scaled, y_train)

# Predict
y_pred_lr = lr.predict(X_test_scaled)

# Evaluate
acc_lr = accuracy_score(y_test, y_pred_lr)
f1_lr = f1_score(y_test, y_pred_lr, average='weighted')

print("Logistic Regression Accuracy:", acc_lr)
print("Logistic Regression F1 Score:", f1_lr)
```

```python
from sklearn.naive_bayes import GaussianNB

# Create model
nb = GaussianNB()

# Train model (NO scaling needed, but scaled also works)
nb.fit(X_train, y_train)

# Predict
y_pred_nb = nb.predict(X_test)

# Evaluate
acc_nb = accuracy_score(y_test, y_pred_nb)
f1_nb = f1_score(y_test, y_pred_nb, average='weighted')

print("Naive Bayes Accuracy:", acc_nb)
print("Naive Bayes F1 Score:", f1_nb)
```

```python
# Print comparison of all models

print("----- Model Comparison -----\n")

print(f"KNN -> Accuracy: {acc_knn:.3f}, F1 Score: {f1_knn:.3f}")
print(f"Logistic Regression -> Accuracy: {acc_lr:.3f}, F1 Score: {f1_lr:.3f}")
print(f"Naive Bayes -> Accuracy: {acc_nb:.3f}, F1 Score: {f1_nb:.3f}")
```
